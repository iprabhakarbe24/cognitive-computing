```python
import random
from collections import Counter

random.seed(1024170037)
```

```python
nums = [random.randint(100,900) for _ in range(100)]
print("generated list of 100 random numbers:")
print(nums)
```

```python
odd_numbers = [n for n in nums if n% 2 ==0]
print(" Odd numbers count:", len(odd_numbers))
print("odd numbers:", odd_numbers)
```

```python
even_numbers = [n for n in nums if n%2 == 0]
print("even numbers count:" , len(even_numbers))
print("even numbers:", even_numbers)
```

```python
def is_prime(n):
    if n< 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n%i==0:
            return False
    return True
prime_numbers = [n for n in nums if is_prime(n)]
print("prime numbers count:", len(prime_numbers))
print("prime numbers:", prime_numbers)
```

```python
counts= Counter(nums)
most_common_nums, most_common_count = counts.most_common(1)[0]
print(f"Most frequent number: {most_common_nums}, occurs {most_common_count} time(s)")
```

```python
roll_number = "1024170037"
digits = [int(d) for d in roll_number[:8]] 
print("Digits used:", digits)
```

```python
A= {digit * 7 for digit in digits}
B={digit * 9 for digit in digits}
print("set A:", A)
print("Set B:", B)
```

```python
union_ab = A.union(B)
print("A union B:", union_ab)
```

```python
intersection_ab = A.intersection(B)
print("A intersection B:", intersection_ab)
```

```python
a_minus_b = A.difference(B)
b_minus_a = B.difference(A)
print(" A-B:", a_minus_b)
print("B-A:", b_minus_a)
```

```python
sym_diff= A.symmetric_difference(B)
print("Symmetric difference(A^B):", sym_diff)
```

```python
is_subset=A.issubset(B)
is_superset=B.issuperset(A)
print("A subset of B?", is_subset)
print("B a superset of A?", is_superset)
```

```python
X= int(input("enter a value X to remove from set A:"))
A.discard(X)
print("Set A after discard():", A)
```
