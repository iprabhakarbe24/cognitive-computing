```python
import pandas as pd
roll_number = "1024170037"

fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee", "category": "billing"},
]
last_two_digits = [int(d) for d in roll_number[-2:]]
categories = ["billing", "account", "general"]

digit1,digit2= last_two_digits
category1= categories[digit1 % 3]
category2= categories[digit2 % 3]

personalized_entry_1 = {
    "question": "how do I get a refund for a duplicate payment",
    "answer": "Duplicate payments are refunded within 5-7 business days after you raise a request in the Billing section.",
    "keywords": "refund duplicate payment",
    "category": category1
}
personalized_entry_2 = {
    "question": "how do I update my registered mobile number",
    "answer": "Go to Profile > Account Settings > Update Mobile Number and verify with OTP.",
    "keywords": "update mobile number",
    "category": category2
}
all_entries = fixed_entries + [personalized_entry_1, personalized_entry_2]
df = pd.DataFrame(all_entries)
print(df.to_string(index=True))
```

```python
#2

def score_query(query,dataframe):
    query_words = set(query.lower(),split())
    matches=[]
    for idx, row in dataframe.iterrows():
        keyword_words = set(row["keywords"].lower().split())
        score = len(query_words & keyword_words)
        if score > 0:
            matches.append({
                "question": row["question"], "answer": row["answer"],
                "cateegory": row["category"], "score": score
            })
    matches.sort(key=lambda x: x["score"], reverse=True)
    return matches
```

```python
#6
def score_query_with_ties(query, dataframe):
    query_words = set(query.lower().split())
    matches = []
    for idx, row in dataframe.iterrows():
        keyword_words = set(row["keywords"].lower().split())
        score = len(query_words & keyword_words)
        if score > 0:
            matches.append({
                "question": row["question"], "answer": row["answer"],
                "category": row["category"], "score": score
            })

    if not matches:
        print("No matching entries found.")
        return []

    matches.sort(key=lambda x: x["score"], reverse=True)
    top_score = matches[0]["score"]
    top_matches = [m for m in matches if m["score"] == top_score]

    if len(top_matches) > 1:
        print(f"TIE DETECTED: {len(top_matches)} entries tie at top score {top_score}:")
        for m in top_matches:
            print(f"  [score {m['score']}] {m['question']} -> {m['answer']} (category: {m['category']})")
    else:
        print(f"Best match [score {top_score}]: {top_matches[0]['question']} -> {top_matches[0]['answer']}")

    return matches
```

```python
score_query_with_ties("fee", df)                             
score_query_with_ties("reset my password login", df)
```
