```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from collections import Counter

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)

text = """David Fincher is known for making dark and carefully planned films.
His movies often explore crime, obsession, and human behavior.
Fight Club became famous for its unusual story and memorable scenes.
The Social Network shows the tense beginnings of Facebook.
Gone Girl keeps viewers guessing until the end.
I like how Fincher uses music, lighting, and small details to build suspense."""


clean_text = re.sub(r"[^a-zA-Z0-9\s]", "", text.lower())
print("Clean text:")
print(clean_text)


print("\nSentences:")
print(sent_tokenize(text))

print("\nPython split():")
print(clean_text.split())

print("\nNLTK word_tokenize():")
print(word_tokenize(clean_text))

print("\nDifference:")
print("split() separates text at whitespace, while word_tokenize() uses language-aware tokenization.")


stop_words = set(stopwords.words("english"))
words = word_tokenize(clean_text)
useful_words = [word for word in words if word.isalpha() and word not in stop_words]

print("\nWords after removing stopwords:")
print(useful_words)


frequency = Counter(useful_words)
print("\nWord frequencies:")
print(frequency.most_common())
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

text = """David Fincher is known for making dark and carefully planned films.
His movies often explore crime, obsession, and human behavior.
Fight Club became famous for its unusual story and memorable scenes.
The Social Network shows the tense beginnings of Facebook.
Gone Girl keeps viewers guessing until the end.
I like how Fincher uses music, lighting, and small details to build suspense."""

words = re.findall(r"[a-zA-Z]+", text.lower())
stop_words = set(stopwords.words("english"))
words = [word for word in words if word not in stop_words]

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed_words = [stemmer.stem(word) for word in words]
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]

print("Words after removing stopwords:")
print(words)

print("\nStemmed words:")
print(stemmed_words)

print("\nLemmatized words:")
print(lemmatized_words)

print("\nSimple comparison:")
print("Stemming cuts words down, so the result may not be a real word.")
print("Lemmatization tries to return a proper dictionary form.")
print("Use stemming for quick text processing and lemmatization when readability matters.")
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "The AI bubble debate is growing as companies spend heavily on artificial intelligence. Some people expect useful tools, while others worry that investment is moving too fast.",
    "The 2008 financial crisis began after risky home loans and financial products caused major losses. Banks failed, jobs disappeared, and many families struggled.",
    "The Social Network tells a dramatized story about the early days of Facebook. It focuses on ambition, friendship, lawsuits, and the cost of building a major technology company."
]


count_vectorizer = CountVectorizer(stop_words="english")
bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words feature names:")
print(count_vectorizer.get_feature_names_out())
print("\nBag of Words matrix:")
print(bow_matrix.toarray())


tfidf_vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)
terms = tfidf_vectorizer.get_feature_names_out()

for i, text in enumerate(texts):
    scores = tfidf_matrix[i].toarray()[0]
    top_indices = scores.argsort()[-3:][::-1]
    top_words = [(terms[index], round(scores[index], 3)) for index in top_indices]
    print(f"\nText {i + 1} top 3 TF-IDF keywords:")
    print(top_words)

print("\nInterpretation:")
print("The AI text highlights terms about AI and investment; the crisis text highlights financial trouble;")
print("the film text highlights the movie and its themes. Exact top words depend on the text.")
```

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

langchain_text = """LangChain helps developers build applications powered by language models.
It connects a model with prompts, documents, tools, and outside data.
It is useful for chatbots and retrieval-based question answering.
Developers can combine different steps into a workflow."""

langgraph_text = """LangGraph helps developers create workflows with language models.
It represents an application as a graph of steps and shared state.
It is useful when an AI task needs loops, decisions, or human approval.
Developers can manage multi-step agents more clearly."""

def clean_words(text):
    return set(re.findall(r"[a-zA-Z]+", text.lower()))

words1 = clean_words(langchain_text)
words2 = clean_words(langgraph_text)

intersection = words1.intersection(words2)
union = words1.union(words2)
jaccard = len(intersection) / len(union) if union else 0

vectorizer = TfidfVectorizer(stop_words="english")
tfidf = vectorizer.fit_transform([langchain_text, langgraph_text])
cosine = cosine_similarity(tfidf[0:1], tfidf[1:2])[0][0]

print("LangChain words:", sorted(words1))
print("\nLangGraph words:", sorted(words2))
print("\nJaccard similarity:", round(jaccard, 3))
print("Cosine similarity:", round(cosine, 3))

print("\nAnalysis:")
print("Jaccard compares which unique words the texts share.")
print("Cosine similarity compares TF-IDF weighted word patterns, so it also considers how important words are.")
print("Here, cosine similarity can give a more useful comparison because the texts discuss related ideas using different details.")
```

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

reviews = [
    "The Sony WH-XB910N headphones have powerful bass and comfortable ear cups. The noise cancelling is useful on buses and during study.",
    "The battery lasts a long time and the sound is enjoyable. The app also gives useful sound settings.",
    "The headphones feel a little bulky, and the microphone is not great in noisy places.",
    "The bass is strong, but sometimes it sounds too heavy for podcasts."
]

positive_reviews = []

for number, review in enumerate(reviews, start=1):
    result = TextBlob(review)
    polarity = result.sentiment.polarity
    subjectivity = result.sentiment.subjectivity

    if polarity > 0.1:
        label = "Positive"
        positive_reviews.append(review)
    elif polarity < -0.1:
        label = "Negative"
    else:
        label = "Neutral"

    print(f"Review {number}: {review}")
    print("Polarity:", round(polarity, 3))
    print("Subjectivity:", round(subjectivity, 3))
    print("Sentiment:", label)
    print()


if positive_reviews:
    all_positive_text = " ".join(positive_reviews)
    cloud = WordCloud(width=800, height=400, background_color="white").generate(all_positive_text)

    plt.figure(figsize=(10, 5))
    plt.imshow(cloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Words from positive headphone reviews")
    plt.show()
else:
    print("No positive reviews were found, so a word cloud was not created.")

print("Note: TextBlob scores are automatic estimates, so they may not perfectly match a human opinion.")
```

```python


import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

text = """Humor has changed a lot over time. In the past, people shared jokes through stories, plays, books, and live performances.
Radio and television helped comedians reach larger audiences. Later, sitcoms and stand up comedy became popular in many homes.
The internet made jokes travel much faster. Memes, short videos, and online trends now shape what many people find funny.
A joke can become popular in one day and feel old the next. Even so, humor still helps people relax, connect, and deal with difficult moments.
What changes is the style, the platform, and the references people understand. Good humor often notices ordinary life and presents it in a surprising way."""

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index) + 1


input_sequences = []
for line in text.split("."):
    sequence = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(sequence)):
        input_sequences.append(sequence[:i + 1])

max_length = max(len(sequence) for sequence in input_sequences)
padded_sequences = pad_sequences(input_sequences, maxlen=max_length, padding="pre")

X = padded_sequences[:, :-1]
y = padded_sequences[:, -1]
y = tf.keras.utils.to_categorical(y, num_classes=total_words)

model = Sequential([
    Embedding(total_words, 16, input_length=max_length - 1),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
model.summary()

#
model.fit(X, y, epochs=80, verbose=0)
print("Model training complete.")

def generate_text(seed_word, next_words=8):
    generated = seed_word.lower()

    for _ in range(next_words):
        sequence = tokenizer.texts_to_sequences([generated])[0]
        sequence = pad_sequences([sequence], maxlen=max_length - 1, padding="pre")
        prediction = model.predict(sequence, verbose=0)[0]
        next_id = int(np.argmax(prediction))

        next_word = ""
        for word, index in tokenizer.word_index.items():
            if index == next_id:
                next_word = word
                break

        if not next_word:
            break
        generated += " " + next_word

    return generated

print("\nGenerated lines:")
print(generate_text("humor", 8))
print(generate_text("people", 8))
print(generate_text("internet", 8))
```
