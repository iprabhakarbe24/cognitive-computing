```python
employees = pd.read_csv(r"D:\cog computing\employees.csv")
print(employees.shape)
```

```python
print(employees.info())
```

```python
print(employees.describe())
```

```python
print(employees.head())
print(employees.tail(3))
```

```python
print(employees["Salary"].mean())
print(employees["Bonus"].sum())
print(employees["Age"].min())
print(employees["Rating"].max())
```

```python
employees = employees.sort_values("Salary",ascending=False)
print(employees)
```

```python
def category(rating):
    if rating>=4.5:
        return "Excellent"
    elif rating >=4.0:
        return "Good"
    else:
        return "Average"
employees["Performance"]= employees["Rating"].apply(category)
print(employees)
```

```python
print(employees.isnull().sum())
```

```python
employees= employees.rename(columns={"Employee_ID":"ID"})
print(employees)
```

```python
print(employees[employees["Years_of_Experience"] > 5])
```

```python
print(employees[employees["Department"]=="IT"])
```

```python
employees["Tax"] = employees["Salary"]*0.10
print(employees)
```
